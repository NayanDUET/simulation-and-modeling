```python
xb = [80, 90, 99, 108, 116, 125, 133, 141, 151, 160, 169, 179, 180]
yb = [0, -2, -5, -9, -15, -18, -23, -29, -28, -25, -21, -20, -17]
n = len(xb)
x_f = [0]; y_f = [50]; v_f = 20; t = 0
```

```python
from math import sqrt
while (t<=n):
    dist = sqrt((y_f[-1] - yb[t])**2 + (x_f[-1] - xb[t])**2)
    print(str(t) + " --> "+ str(dist) )
    if(dist <= 10.0):
        print(t, dist, "Finished")
        break
    else: t+=1

    sin_th = ( yb[t] - y_f[-1] )/dist
    cos_th = ( xb[t] - x_f[-1] )/dist

    x_f.append(x_f[-1] + v_f*cos_th)
    y_f.append(y_f[-1] + v_f*sin_th)

if(t>n):
    print("Target Escaped.")    
```

```python
import matplotlib.pyplot as plt 
```

```python
plt.plot(xb, yb, label='Bomber path')
plt.scatter(xb, yb, label='Bomber points')

plt.plot(x_f, y_f, label='Fighter path')
plt.scatter(x_f, y_f, label='Fighter points')

plt.scatter(x_f[-1], y_f[-1], color='red')
plt.text(x_f[-1], y_f[-1], f'({round(x_f[-1], 2)}, {round(y_f[-1])})')
plt.text(xb[t], yb[t], f'({round(xb[t], 2)}, {round(yb[t])})', horizontalalignment='right')

plt.legend()
plt.show()
```
