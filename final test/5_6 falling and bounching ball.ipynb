```python
import numpy as np
import matplotlib.pyplot as plt

# Given values
H = 100        # initial height (m)
K = 0.6        # coefficient of restitution
g = 9.8        # gravity (m/s^2)
bounces = 30   # number of bounces

time = []
height = []

t_current = 0

# Initial fall
t_fall = np.sqrt(2 * H / g)
time.append(t_current)
height.append(H)

time.append(t_current + t_fall)
height.append(0)

t_current += t_fall
h_current = H

# Bounces
for n in range(1, bounces + 1):
    h_current = h_current * K**2
    t_up = np.sqrt(2 * h_current / g)
    
    # Upward motion
    time.append(t_current + t_up)
    height.append(h_current)
    
    # Downward motion
    time.append(t_current + 2 * t_up)
    height.append(0)
    
    t_current += 2 * t_up

# Plot
plt.figure()
plt.plot(time, height)
plt.xlabel("Time (s)")
plt.ylabel("Height (m)")
plt.title("Time vs Height of Bouncing Ball")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Given values
H = 100
K = 0.6
g = 9.8
bounces = 20

n = np.arange(0, bounces + 1)

# Impact speed formula
v = np.sqrt(2 * g * H * (K ** (2 * n)))

# Plot
plt.figure()
plt.plot(n, v, marker='o')
plt.xlabel("Bounce Number (n)")
plt.ylabel("Impact Speed (m/s)")
plt.title("Bounce Number vs Impact Speed")
plt.grid(True)
plt.show()
```
