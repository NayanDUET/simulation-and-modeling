```python
# Qi = 600; Qm = -3000; Si = -100; Sm = 2000
Qi = int(input("Enter the intercept of demand: "))
Qm = int(input("Enter the slope of demand: "))
Si = int(input("Enter the intercept of supply: "))
Sm = int(input("Enter the slope of supply: "))

P = (Qi +  (- Si) ) / ((-Qm) + Sm);    # equilibrium price
Qd = Qi + Qm * P;   # equilibrium quantity
Qs = Si + Sm * P;

print("Simple Market Model (Static)\n");
print("-----------------------------------\n");
print("Equilibrium Price (P) = ", P);
print("Equilibrium Quantity (Qd) = ", Qd);
print("Equilibrium Quantity (Qs) = ", Qs);
```

```python
qd = []
qs = []
prices = []
p = 0.0
while p <= 0.3:
    qd.append(600 - 3000 * p)
    qs.append(-100 + 2000 * p)
    prices.append(p)
    p += 0.05

```

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(prices, qd, label="Demand (Qd)")
plt.plot(prices, qs, label="Supply (Qs)")
plt.scatter(P, Qd, color='red')
plt.text(P, Qd, f"  (P*={round(P,2)}, Q*={round(Qd,2)})")
plt.xlabel("Price (P)")
plt.ylabel("Quantity")
plt.title("Supply and Demand Curves")
plt.legend()
plt.grid(True)
plt.show()
```
